# 1. **Start and Run Code**

Google Colab is a browser-based notebook for running Python code.

A notebook contains **code cells** and **text cells**.

To start working:

1. Open the notebook.
2. Click **Connect** if Colab is not connected.
3. Click the ▶ button beside a code cell, or press `Shift + Enter`.
4. Read the output below the cell.


````python
print("Hello Colab!")
````

The code above prints text below the cell.

You can edit the code, run it again, and immediately see the new result.


# 2. **Execution Order**

Cells can depend on values created by earlier cells.

Run this first:


````python
x = 10
````

Then run this:


````python
print(x)
````

If you restart the runtime, stored variables such as `x` are cleared.

After a restart, run the required cells again from the beginning.

If a cell is taking too long, use the **Stop** button to stop it.


# 3. **Files in Colab**

Colab uses a temporary Linux file system.

The main working folder is usually:

```text
/content
```

Use these commands to see where you are and what files are available.


````python
!pwd
!ls
````

- `!pwd` shows the current folder.
- `!ls` lists files and folders.
- `!` lets you run a Linux command inside a code cell.

To change the working folder, use `%cd`.


````python
%cd /content
````

Files stored only in `/content` are temporary and may disappear when the runtime ends.
